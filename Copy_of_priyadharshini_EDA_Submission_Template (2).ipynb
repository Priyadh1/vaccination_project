````python
# ============================================================
# STEP 3.2 — REMOVE CONFIRMED METADATA / FOOTER ROWS
# ============================================================

clean_datasets = {
    "Coverage": coverage_clean,
    "Incidence Rate": incidence_clean,
    "Reported Cases": cases_clean,
    "Vaccine Introduction": introduction_clean,
    "Vaccine Schedule": schedule_clean
}

for name, df in clean_datasets.items():

    # Detect rows containing source metadata such as:
    # "Created: 2025-02-01 16:00 UTC"
    metadata_mask = df.astype(str).apply(
        lambda column: column.str.contains(
            r"^Created:",
            case=False,
            na=False
        )
    ).any(axis=1)

    # Remove only confirmed metadata rows
    clean_datasets[name] = df.loc[~metadata_mask].copy()

# Update the cleaned DataFrames
coverage_clean = clean_datasets["Coverage"]
incidence_clean = clean_datasets["Incidence Rate"]
cases_clean = clean_datasets["Reported Cases"]
introduction_clean = clean_datasets["Vaccine Introduction"]
schedule_clean = clean_datasets["Vaccine Schedule"]

print("✓ Confirmed metadata/footer rows removed.")
print()
print("Cleaned dataset dimensions:")
print(f"Coverage:             {coverage_clean.shape}")
print(f"Incidence Rate:       {incidence_clean.shape}")
print(f"Reported Cases:       {cases_clean.shape}")
print(f"Vaccine Introduction: {introduction_clean.shape}")
print(f"Vaccine Schedule:     {schedule_clean.shape}")
````

````python
# ============================================================
# STEP 3.8 — HANDLE INVALID DOSE VALUES
# ============================================================

# Preserve the original dose values for audit purposes
coverage_clean["DOSES_RAW"] = coverage_clean["DOSES"]

# Create an explicit quality flag
coverage_clean["INVALID_DOSES_FLAG"] = (
    coverage_clean["DOSES"] < 0
)

# Replace only confirmed negative dose values with NaN
coverage_clean.loc[
    coverage_clean["INVALID_DOSES_FLAG"],
    "DOSES"
] = np.nan

print("✓ Invalid dose values handled.")
print()

print(
    "Invalid dose records flagged:",
    coverage_clean["INVALID_DOSES_FLAG"].sum()
)

print(
    "Negative values remaining in cleaned DOSES:",
    (coverage_clean["DOSES"] < 0).sum()
)

print()
print("The original values are preserved in DOSES_RAW.")
````

````python
# ============================================================
# STEP 3.9 — FLAG COVERAGE VALUES ABOVE 100%
# ============================================================

# Create an explicit anomaly flag
coverage_clean["COVERAGE_ABOVE_100_FLAG"] = (
    coverage_clean["COVERAGE"] > 100
)

print("✓ Coverage anomaly flag created.")
print()

print(
    "Coverage records above 100%:",
    coverage_clean["COVERAGE_ABOVE_100_FLAG"].sum()
)

print(
    "Coverage records at or below 100%:",
    (~coverage_clean["COVERAGE_ABOVE_100_FLAG"]).sum()
)

print()
print("Original COVERAGE values have NOT been modified.")
````

````python
# ============================================================
# STEP 3.10 — VALIDATE YEAR VALUES
# ============================================================

print("YEAR QUALITY AUDIT")
print("=" * 80)

for name, df in clean_datasets.items():

    year_series = df["YEAR"].dropna()

    # Check for fractional years
    fractional_years = year_series[
        year_series % 1 != 0
    ]

    # Check for unrealistic years
    invalid_years = year_series[
        (year_series < 1900) | (year_series > 2023)
    ]

    print(f"\n{name}")
    print("-" * 80)
    print(f"Missing YEAR values:       {df['YEAR'].isna().sum():,}")
    print(f"Fractional YEAR values:    {len(fractional_years):,}")
    print(f"Out-of-range YEAR values:  {len(invalid_years):,}")

print("\n" + "=" * 80)
print("✓ Year-quality audit completed.")
````

````python
# ============================================================
# STEP 3.11 — POST-CLEANING MISSING VALUE AUDIT
# ============================================================

print("POST-CLEANING MISSING VALUE AUDIT")
print("=" * 90)

for name, df in clean_datasets.items():

    missing_count = df.isna().sum()
    missing_percent = (missing_count / len(df)) * 100

    audit = pd.DataFrame({
        "Missing_Count": missing_count,
        "Missing_Percentage": missing_percent.round(2)
    })

    audit = audit[audit["Missing_Count"] > 0]

    print("\n" + "=" * 90)
    print(name.upper())
    print("=" * 90)

    if len(audit) > 0:
        display(audit)
    else:
        print("No missing values found.")

print("\n" + "=" * 90)
print("✓ Post-cleaning missing-value audit completed.")
````

````python
# ============================================================
# STEP 3.12 — ANALYZE COVERAGE MISSINGNESS BY CATEGORY
# ============================================================

coverage_missing_by_category = (
    coverage_clean
    .groupby("COVERAGE_CATEGORY")
    .agg(
        Total_Records=("COVERAGE", "size"),
        Missing_Coverage=("COVERAGE", lambda x: x.isna().sum()),
        Missing_Doses=("DOSES", lambda x: x.isna().sum()),
        Missing_Target=("TARGET_NUMBER", lambda x: x.isna().sum())
    )
)

coverage_missing_by_category["Coverage_Missing_%"] = (
    coverage_missing_by_category["Missing_Coverage"]
    / coverage_missing_by_category["Total_Records"]
    * 100
).round(2)

display(coverage_missing_by_category)

print("\n✓ Coverage missingness by category calculated.")
````

````python
# ============================================================
# SQL DATABASE PREPARATION
# STEP 3.11 — CREATE SQL DATABASE CONNECTION
# ============================================================

DB_PATH = PROJECT_DIR / "vaccination_database.db"

conn = sqlite3.connect(DB_PATH)

print("✓ SQL database connection created successfully.")
print(f"Database location: {DB_PATH}")
````

````python
# ============================================================
# STEP 3.12 — CREATE NORMALIZED SQL TABLES
# ============================================================

cursor = conn.cursor()

# ------------------------------------------------------------
# 1. COUNTRY DIMENSION
# ------------------------------------------------------------
cursor.execute("""
CREATE TABLE IF NOT EXISTS dim_country (
    country_id INTEGER PRIMARY KEY AUTOINCREMENT,
    iso_3_code TEXT UNIQUE,
    country_name TEXT,
    who_region TEXT
)
""")

# ------------------------------------------------------------
# 2. VACCINE DIMENSION
# ------------------------------------------------------------
cursor.execute("""
CREATE TABLE IF NOT EXISTS dim_vaccine (
    vaccine_id INTEGER PRIMARY KEY AUTOINCREMENT,
    vaccine_code TEXT UNIQUE,
    vaccine_description TEXT
)
""")

# ------------------------------------------------------------
# 3. DISEASE DIMENSION
# ------------------------------------------------------------
cursor.execute("""
CREATE TABLE IF NOT EXISTS dim_disease (
    disease_id INTEGER PRIMARY KEY AUTOINCREMENT,
    disease_code TEXT UNIQUE,
    disease_description TEXT
)
""")

# ------------------------------------------------------------
# 4. YEAR DIMENSION
# ------------------------------------------------------------
cursor.execute("""
CREATE TABLE IF NOT EXISTS dim_year (
    year_id INTEGER PRIMARY KEY,
    year INTEGER UNIQUE
)
""")

# ------------------------------------------------------------
# 5. COVERAGE FACT TABLE
# ------------------------------------------------------------
cursor.execute("""
CREATE TABLE IF NOT EXISTS fact_coverage (
    coverage_id INTEGER PRIMARY KEY AUTOINCREMENT,
    country_id INTEGER,
    year_id INTEGER,
    antigen TEXT,
    antigen_description TEXT,
    coverage_category TEXT,
    coverage_category_description TEXT,
    target_number REAL,
    doses REAL,
    doses_raw REAL,
    coverage REAL,
    invalid_doses_flag INTEGER,
    coverage_above_100_flag INTEGER,

    FOREIGN KEY (country_id) REFERENCES dim_country(country_id),
    FOREIGN KEY (year_id) REFERENCES dim_year(year_id)
)
""")

# ------------------------------------------------------------
# 6. INCIDENCE FACT TABLE
# ------------------------------------------------------------
cursor.execute("""
CREATE TABLE IF NOT EXISTS fact_incidence (
    incidence_id INTEGER PRIMARY KEY AUTOINCREMENT,
    country_id INTEGER,
    year_id INTEGER,
    disease_id INTEGER,
    denominator TEXT,
    incidence_rate REAL,

    FOREIGN KEY (country_id) REFERENCES dim_country(country_id),
    FOREIGN KEY (year_id) REFERENCES dim_year(year_id),
    FOREIGN KEY (disease_id) REFERENCES dim_disease(disease_id)
)
""")

# ------------------------------------------------------------
# 7. REPORTED CASES FACT TABLE
# ------------------------------------------------------------
cursor.execute("""
CREATE TABLE IF NOT EXISTS fact_cases (
    cases_id INTEGER PRIMARY KEY AUTOINCREMENT,
    country_id INTEGER,
    year_id INTEGER,
    disease_id INTEGER,
    cases REAL,

    FOREIGN KEY (country_id) REFERENCES dim_country(country_id),
    FOREIGN KEY (year_id) REFERENCES dim_year(year_id),
    FOREIGN KEY (disease_id) REFERENCES dim_disease(disease_id)
)
""")

# ------------------------------------------------------------
# 8. VACCINE INTRODUCTION FACT TABLE
# ------------------------------------------------------------
cursor.execute("""
CREATE TABLE IF NOT EXISTS fact_vaccine_introduction (
    introduction_id INTEGER PRIMARY KEY AUTOINCREMENT,
    country_id INTEGER,
    year_id INTEGER,
    description TEXT,
    intro_status TEXT,

    FOREIGN KEY (country_id) REFERENCES dim_country(country_id),
    FOREIGN KEY (year_id) REFERENCES dim_year(year_id)
)
""")

# ------------------------------------------------------------
# 9. VACCINE SCHEDULE FACT TABLE
# ------------------------------------------------------------
cursor.execute("""
CREATE TABLE IF NOT EXISTS fact_vaccine_schedule (
    schedule_id INTEGER PRIMARY KEY AUTOINCREMENT,
    country_id INTEGER,
    year_id INTEGER,
    vaccine_id INTEGER,
    schedule_rounds REAL,
    target_pop TEXT,
    target_pop_description TEXT,
    geoarea TEXT,
    age_administered TEXT,
    source_comment TEXT,

    FOREIGN KEY (country_id) REFERENCES dim_country(country_id),
    FOREIGN KEY (year_id) REFERENCES dim_year(year_id),
    FOREIGN KEY (vaccine_id) REFERENCES dim_vaccine(vaccine_id)
)
""")

conn.commit()

print("✓ All normalized SQL tables created successfully.")
````

````python
# ============================================================
# STEP 3.13 — VIEW SQL TABLES
# ============================================================

tables = pd.read_sql_query("""
    SELECT name
    FROM sqlite_master
    WHERE type = 'table'
    ORDER BY name;
""", conn)

display(tables)

print(f"\n✓ Total tables created: {len(tables)}")
````

````python
# ============================================================
# STEP 3.14 — POPULATE SQL DIMENSION TABLES
# SAFE / RE-RUNNABLE VERSION
# ============================================================

# ------------------------------------------------------------
# 1. DIM_COUNTRY
# ------------------------------------------------------------

country_parts = []

country_parts.append(
    coverage_df.loc[
        coverage_df["GROUP"].eq("COUNTRIES"),
        ["CODE", "NAME"]
    ].rename(columns={
        "CODE": "ISO_3_CODE",
        "NAME": "COUNTRY_NAME"
    })
)

country_parts.append(
    introduction_df[
        ["ISO_3_CODE", "COUNTRYNAME", "WHO_REGION"]
    ].rename(columns={
        "COUNTRYNAME": "COUNTRY_NAME"
    })
)

country_parts.append(
    schedule_df[
        ["ISO_3_CODE", "COUNTRYNAME", "WHO_REGION"]
    ].rename(columns={
        "COUNTRYNAME": "COUNTRY_NAME"
    })
)

country_df = pd.concat(country_parts, ignore_index=True)

country_df["ISO_3_CODE"] = (
    country_df["ISO_3_CODE"]
    .astype("string")
    .str.strip()
)

country_df["COUNTRY_NAME"] = (
    country_df["COUNTRY_NAME"]
    .astype("string")
    .str.strip()
)

country_df["WHO_REGION"] = (
    country_df["WHO_REGION"]
    .astype("string")
    .str.strip()
)

country_df = (
    country_df
    .dropna(subset=["ISO_3_CODE"])
    .drop_duplicates(subset=["ISO_3_CODE"])
    .reset_index(drop=True)
)

country_sql = country_df[
    ["ISO_3_CODE", "COUNTRY_NAME", "WHO_REGION"]
]

# Check countries already present in database
existing_countries = pd.read_sql_query(
    "SELECT iso_3_code FROM dim_country",
    conn
)

country_sql = country_sql[
    ~country_sql["ISO_3_CODE"].isin(
        existing_countries["iso_3_code"]
    )
]

if not country_sql.empty:
    country_sql.to_sql(
        "dim_country",
        conn,
        if_exists="append",
        index=False
    )


# ------------------------------------------------------------
# 2. DIM_DISEASE
# ------------------------------------------------------------

disease_df = pd.concat([
    incidence_df[
        ["DISEASE", "DISEASE_DESCRIPTION"]
    ],
    cases_df[
        ["DISEASE", "DISEASE_DESCRIPTION"]
    ]
], ignore_index=True)

disease_df = (
    disease_df
    .dropna(subset=["DISEASE"])
    .drop_duplicates(subset=["DISEASE"])
    .rename(columns={
        "DISEASE": "DISEASE_CODE"
    })
    .reset_index(drop=True)
)

# Check diseases already present in database
existing_diseases = pd.read_sql_query(
    "SELECT disease_code FROM dim_disease",
    conn
)

disease_df = disease_df[
    ~disease_df["DISEASE_CODE"].isin(
        existing_diseases["disease_code"]
    )
]

if not disease_df.empty:
    disease_df.to_sql(
        "dim_disease",
        conn,
        if_exists="append",
        index=False
    )


# ------------------------------------------------------------
# 3. DIM_VACCINE
# ------------------------------------------------------------

vaccine_df = (
    schedule_df[
        ["VACCINECODE", "VACCINE_DESCRIPTION"]
    ]
    .dropna(subset=["VACCINECODE"])
    .drop_duplicates(subset=["VACCINECODE"])
    .rename(columns={
        "VACCINECODE": "VACCINE_CODE"
    })
    .reset_index(drop=True)
)

# Check vaccines already present in database
existing_vaccines = pd.read_sql_query(
    "SELECT vaccine_code FROM dim_vaccine",
    conn
)

vaccine_df = vaccine_df[
    ~vaccine_df["VACCINE_CODE"].isin(
        existing_vaccines["vaccine_code"]
    )
]

if not vaccine_df.empty:
    vaccine_df.to_sql(
        "dim_vaccine",
        conn,
        if_exists="append",
        index=False
    )


# ------------------------------------------------------------
# 4. DIM_YEAR
# ------------------------------------------------------------

all_years = pd.concat([
    coverage_df["YEAR"],
    incidence_df["YEAR"],
    cases_df["YEAR"],
    introduction_df["YEAR"],
    schedule_df["YEAR"]
]).dropna()

year_values = sorted(
    all_years.astype(int).unique()
)

year_df = pd.DataFrame({
    "year_id": year_values,
    "year": year_values
})

# Check years already present in database
existing_years = pd.read_sql_query(
    "SELECT year_id FROM dim_year",
    conn
)

year_df = year_df[
    ~year_df["year_id"].isin(
        existing_years["year_id"]
    )
]

if not year_df.empty:
    year_df.to_sql(
        "dim_year",
        conn,
        if_exists="append",
        index=False
    )


# ------------------------------------------------------------
# 5. COMMIT
# ------------------------------------------------------------

conn.commit()


# ------------------------------------------------------------
# 6. FINAL COUNTS
# ------------------------------------------------------------

final_country_count = pd.read_sql_query(
    "SELECT COUNT(*) AS n FROM dim_country",
    conn
).iloc[0]["n"]

final_disease_count = pd.read_sql_query(
    "SELECT COUNT(*) AS n FROM dim_disease",
    conn
).iloc[0]["n"]

final_vaccine_count = pd.read_sql_query(
    "SELECT COUNT(*) AS n FROM dim_vaccine",
    conn
).iloc[0]["n"]

final_year_count = pd.read_sql_query(
    "SELECT COUNT(*) AS n FROM dim_year",
    conn
).iloc[0]["n"]


print("✓ Dimension tables populated successfully.")
print()
print(f"Countries : {final_country_count:,}")
print(f"Diseases  : {final_disease_count:,}")
print(f"Vaccines  : {final_vaccine_count:,}")
print(f"Years     : {final_year_count:,}")
````

````python
# ============================================================
# STEP 3.17 — LOAD COVERAGE DATA INTO SQL
# ============================================================

coverage_sql = coverage_df.copy()

# Map country codes to country_id
coverage_sql = coverage_sql.merge(
    country_map[["country_id", "iso_3_code"]],
    left_on="CODE",
    right_on="iso_3_code",
    how="left"
)

# Map years to year_id
coverage_sql = coverage_sql.merge(
    year_map[["year_id", "year"]],
    left_on="YEAR",
    right_on="year",
    how="left"
)

# Prepare columns according to the SQL table
coverage_sql = coverage_sql[
    [
        "country_id",
        "year_id",
        "ANTIGEN",
        "ANTIGEN_DESCRIPTION",
        "COVERAGE_CATEGORY",
        "COVERAGE_CATEGORY_DESCRIPTION",
        "TARGET_NUMBER",
        "DOSES",
        "DOSES_RAW",
        "COVERAGE",
        "INVALID_DOSES_FLAG",
        "COVERAGE_ABOVE_100_FLAG"
    ]
].copy()

# Convert Boolean flags to SQLite-compatible integers
coverage_sql["INVALID_DOSES_FLAG"] = (
    coverage_sql["INVALID_DOSES_FLAG"].astype(int)
)

coverage_sql["COVERAGE_ABOVE_100_FLAG"] = (
    coverage_sql["COVERAGE_ABOVE_100_FLAG"].astype(int)
)

# Insert into SQL
coverage_sql.to_sql(
    "fact_coverage",
    conn,
    if_exists="append",
    index=False
)

conn.commit()

print("✓ Coverage data loaded into fact_coverage successfully.")
print(f"Rows loaded: {len(coverage_sql):,}")
````

````python
# ============================================================
# STEP 3.19 — LOAD REPORTED CASES DATA INTO SQL
# ============================================================

cases_sql = cases_df.copy()

# Map country codes to country_id
cases_sql = cases_sql.merge(
    country_map[["country_id", "iso_3_code"]],
    left_on="CODE",
    right_on="iso_3_code",
    how="left"
)

# Map years to year_id
cases_sql = cases_sql.merge(
    year_map[["year_id", "year"]],
    left_on="YEAR",
    right_on="year",
    how="left"
)

# Map disease codes to disease_id
cases_sql = cases_sql.merge(
    disease_map,
    left_on="DISEASE",
    right_on="disease_code",
    how="left"
)

# Prepare columns according to SQL table
cases_sql = cases_sql[
    [
        "country_id",
        "year_id",
        "disease_id",
        "CASES"
    ]
].copy()

# Insert into SQL
cases_sql.to_sql(
    "fact_cases",
    conn,
    if_exists="append",
    index=False
)

conn.commit()

print("✓ Reported cases data loaded into fact_cases successfully.")
print(f"Rows loaded: {len(cases_sql):,}")
````

````python
# ============================================================
# STEP 3.27 — PRESERVE SOURCE GROUP IN SQL FACT TABLES
# ============================================================

# Foreign-key enforcement
conn.execute("PRAGMA foreign_keys = ON")

# ------------------------------------------------------------
# 1. Recreate FACT_COVERAGE with SOURCE_GROUP
# ------------------------------------------------------------

conn.execute("DROP TABLE IF EXISTS fact_coverage")

conn.execute("""
CREATE TABLE fact_coverage (
    coverage_id INTEGER PRIMARY KEY AUTOINCREMENT,
    country_id INTEGER,
    year_id INTEGER,
    source_group TEXT,
    antigen TEXT,
    antigen_description TEXT,
    coverage_category TEXT,
    coverage_category_description TEXT,
    target_number REAL,
    doses REAL,
    doses_raw REAL,
    coverage REAL,
    invalid_doses_flag INTEGER,
    coverage_above_100_flag INTEGER,
    FOREIGN KEY (country_id) REFERENCES dim_country(country_id),
    FOREIGN KEY (year_id) REFERENCES dim_year(year_id)
)
""")

coverage_sql = coverage_df.copy()

coverage_sql = coverage_sql.merge(
    country_map[["country_id", "iso_3_code"]],
    left_on="CODE",
    right_on="iso_3_code",
    how="left"
)

coverage_sql = coverage_sql.merge(
    year_map[["year_id", "year"]],
    left_on="YEAR",
    right_on="year",
    how="left"
)

coverage_sql = coverage_sql[
    [
        "country_id",
        "year_id",
        "GROUP",
        "ANTIGEN",
        "ANTIGEN_DESCRIPTION",
        "COVERAGE_CATEGORY",
        "COVERAGE_CATEGORY_DESCRIPTION",
        "TARGET_NUMBER",
        "DOSES",
        "DOSES_RAW",
        "COVERAGE",
        "INVALID_DOSES_FLAG",
        "COVERAGE_ABOVE_100_FLAG"
    ]
].rename(columns={"GROUP": "source_group"})

coverage_sql["invalid_doses_flag"] = (
    coverage_sql["INVALID_DOSES_FLAG"].fillna(False).astype(int)
)

coverage_sql["coverage_above_100_flag"] = (
    coverage_sql["COVERAGE_ABOVE_100_FLAG"].fillna(False).astype(int)
)

coverage_sql.to_sql(
    "fact_coverage",
    conn,
    if_exists="append",
    index=False
)

# ------------------------------------------------------------
# 2. Recreate FACT_INCIDENCE with SOURCE_GROUP
# ------------------------------------------------------------

conn.execute("DROP TABLE IF EXISTS fact_incidence")

conn.execute("""
CREATE TABLE fact_incidence (
    incidence_id INTEGER PRIMARY KEY AUTOINCREMENT,
    country_id INTEGER,
    year_id INTEGER,
    source_group TEXT,
    disease_id INTEGER,
    denominator TEXT,
    incidence_rate REAL,
    FOREIGN KEY (country_id) REFERENCES dim_country(country_id),
    FOREIGN KEY (year_id) REFERENCES dim_year(year_id),
    FOREIGN KEY (disease_id) REFERENCES dim_disease(disease_id)
)
""")

incidence_sql = incidence_df.copy()

incidence_sql = incidence_sql.merge(
    country_map[["country_id", "iso_3_code"]],
    left_on="CODE",
    right_on="iso_3_code",
    how="left"
)

incidence_sql = incidence_sql.merge(
    year_map[["year_id", "year"]],
    left_on="YEAR",
    right_on="year",
    how="left"
)

disease_map = pd.read_sql_query(
    """
    SELECT disease_id, disease_code
    FROM dim_disease
    """,
    conn
)

incidence_sql = incidence_sql.merge(
    disease_map,
    left_on="DISEASE",
    right_on="disease_code",
    how="left"
)

incidence_sql = incidence_sql[
    [
        "country_id",
        "year_id",
        "GROUP",
        "disease_id",
        "DENOMINATOR",
        "INCIDENCE_RATE"
    ]
].rename(columns={
    "GROUP": "source_group",
    "DENOMINATOR": "denominator",
    "INCIDENCE_RATE": "incidence_rate"
})

incidence_sql.to_sql(
    "fact_incidence",
    conn,
    if_exists="append",
    index=False
)

# ------------------------------------------------------------
# 3. Recreate FACT_CASES with SOURCE_GROUP
# ------------------------------------------------------------

conn.execute("DROP TABLE IF EXISTS fact_cases")

conn.execute("""
CREATE TABLE fact_cases (
    cases_id INTEGER PRIMARY KEY AUTOINCREMENT,
    country_id INTEGER,
    year_id INTEGER,
    source_group TEXT,
    disease_id INTEGER,
    cases REAL,
    FOREIGN KEY (country_id) REFERENCES dim_country(country_id),
    FOREIGN KEY (year_id) REFERENCES dim_year(year_id),
    FOREIGN KEY (disease_id) REFERENCES dim_disease(disease_id)
)
""")

cases_sql = cases_df.copy()

cases_sql = cases_sql.merge(
    country_map[["country_id", "iso_3_code"]],
    left_on="CODE",
    right_on="iso_3_code",
    how="left"
)

cases_sql = cases_sql.merge(
    year_map[["year_id", "year"]],
    left_on="YEAR",
    right_on="year",
    how="left"
)

cases_sql = cases_sql.merge(
    disease_map,
    left_on="DISEASE",
    right_on="disease_code",
    how="left"
)

cases_sql = cases_sql[
    [
        "country_id",
        "year_id",
        "GROUP",
        "disease_id",
        "CASES"
    ]
].rename(columns={
    "GROUP": "source_group",
    "CASES": "cases"
})

cases_sql.to_sql(
    "fact_cases",
    conn,
    if_exists="append",
    index=False
)

conn.commit()

# ------------------------------------------------------------
# 4. Basic verification
# ------------------------------------------------------------

print("=" * 60)
print("SOURCE GROUP PRESERVATION")
print("=" * 60)

for table in ["fact_coverage", "fact_incidence", "fact_cases"]:

    count = pd.read_sql_query(
        f"SELECT COUNT(*) AS total_rows FROM {table}",
        conn
    ).iloc[0, 0]

    groups = pd.read_sql_query(
        f"""
        SELECT source_group, COUNT(*) AS rows
        FROM {table}
        GROUP BY source_group
        ORDER BY rows DESC
        """,
        conn
    )

    print(f"\n{table}: {count:,} rows")
    display(groups)
````

````python
# ============================================================
# STEP 4.8A — SCIENTIFICALLY VALID CORRELATION ANALYSIS
# ============================================================

print("=" * 75)
print("4.8A — SCIENTIFICALLY VALID CORRELATION ANALYSIS")
print("=" * 75)

# ------------------------------------------------------------
# Purpose:
# Compare vaccination coverage with MEASLES INCIDENCE,
# rather than raw reported case counts.
#
# Incidence is used because it is already expressed relative
# to a population denominator in the source data.
#
# Only one denominator is selected for the comparison.
# No different incidence denominators are mixed.
# ------------------------------------------------------------

# Find available measles-related vaccination antigens
measles_antigens = (
    coverage_eda[
        coverage_eda["antigen_description"]
        .str.contains("measles", case=False, na=False)
    ][
        ["antigen", "antigen_description"]
    ]
    .drop_duplicates()
    .sort_values("antigen")
)

print("\nAvailable measles-related vaccination antigens:")
display(measles_antigens)


# ------------------------------------------------------------
# Select MCV1 when available.
# Otherwise use the first available measles antigen.
# ------------------------------------------------------------

available_antigens = measles_antigens["antigen"].dropna().tolist()

if "MCV1" in available_antigens:
    selected_antigen = "MCV1"
elif len(available_antigens) > 0:
    selected_antigen = available_antigens[0]
else:
    selected_antigen = None

print(f"\nSelected vaccination antigen: {selected_antigen}")


# ------------------------------------------------------------
# Select WUENIC coverage where available.
# ------------------------------------------------------------

if selected_antigen is not None:

    measles_coverage = coverage_eda[
        (coverage_eda["antigen"] == selected_antigen) &
        (coverage_eda["coverage_category"] == "WUENIC") &
        (coverage_eda["coverage"] >= 0) &
        (coverage_eda["coverage"] <= 100)
    ][
        ["iso_3_code", "year", "coverage"]
    ].dropna()

else:
    measles_coverage = pd.DataFrame(
        columns=["iso_3_code", "year", "coverage"]
    )


# ------------------------------------------------------------
# Measles incidence
# ------------------------------------------------------------

measles_incidence = incidence_eda[
    incidence_eda["disease_code"] == "MEASLES"
].dropna(
    subset=["incidence_rate", "denominator"]
).copy()

print("\nAvailable measles incidence denominators:")
display(
    measles_incidence["denominator"]
    .value_counts()
    .rename_axis("denominator")
    .reset_index(name="records")
)


# ------------------------------------------------------------
# Select the most frequently reported measles denominator.
# This prevents mixing incomparable incidence rates.
# ------------------------------------------------------------

if len(measles_incidence) > 0:

    selected_denominator = (
        measles_incidence["denominator"]
        .value_counts()
        .index[0]
    )

    measles_incidence = measles_incidence[
        measles_incidence["denominator"] == selected_denominator
    ]

else:
    selected_denominator = None


print(
    f"\nSelected measles incidence denominator: "
    f"{selected_denominator}"
)


# ------------------------------------------------------------
# Create country-year comparison dataset
# ------------------------------------------------------------

if selected_antigen is not None and selected_denominator is not None:

    correlation_measles = measles_coverage.merge(
        measles_incidence[
            ["iso_3_code", "year", "incidence_rate"]
        ],
        on=["iso_3_code", "year"],
        how="inner"
    )

else:

    correlation_measles = pd.DataFrame(
        columns=[
            "iso_3_code",
            "year",
            "coverage",
            "incidence_rate"
        ]
    )


print(
    f"\nPaired country-year observations: "
    f"{len(correlation_measles):,}"
)


# ------------------------------------------------------------
# Calculate correlation
# ------------------------------------------------------------

if len(correlation_measles) >= 2:

    pearson_correlation = correlation_measles[
        ["coverage", "incidence_rate"]
    ].corr().iloc[0, 1]

    spearman_correlation = correlation_measles[
        ["coverage", "incidence_rate"]
    ].corr(method="spearman").iloc[0, 1]

    print(
        f"\nPearson correlation : {pearson_correlation:.4f}"
    )

    print(
        f"Spearman correlation: {spearman_correlation:.4f}"
    )

    # --------------------------------------------------------
    # Scatter plot
    # --------------------------------------------------------

    plot_data = correlation_measles.copy()

    # Limit plotting size only for visualization.
    # The correlation above uses the complete paired dataset.
    if len(plot_data) > 5000:
        plot_data = plot_data.sample(
            5000,
            random_state=42
        )

    plt.figure(figsize=(10, 7))

    plt.scatter(
        plot_data["coverage"],
        plot_data["incidence_rate"],
        alpha=0.35
    )

    plt.title(
        "Measles Vaccination Coverage vs Measles Incidence"
    )

    plt.xlabel(
        f"{selected_antigen} Coverage (%)"
    )

    plt.ylabel(
        f"Measles Incidence ({selected_denominator})"
    )

    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

else:

    print(
        "\nNot enough paired observations for a valid correlation."
    )


# ------------------------------------------------------------
# Final interpretation note
# ------------------------------------------------------------

print("\n" + "-" * 75)
print("CORRELATION INTERPRETATION NOTE")
print("-" * 75)

print(
    "This analysis describes the statistical association between "
    "vaccination coverage and measles incidence in the available "
    "country-year observations."
)

print(
    "It does NOT establish causation and should not be interpreted "
    "as proof that changes in vaccination coverage directly caused "
    "changes in incidence."
)

print(
    "\n✓ Only one measles incidence denominator was used."
)

print(
    "✓ Coverage values above 100% were excluded from this "
    "specific correlation calculation."
)

print(
    "✓ No raw case counts were mixed with incidence rates."
)
````
